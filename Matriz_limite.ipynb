# Matriz limite: Estados estacionarios

Una cadena de Markov tiene matriz de transición:

$$\begin{pmatrix}
    0 & 1/4 & 0 & 0 & 3/4 \\
    3/4 & 0 & 0 & 0 & 1/4 \\
    0 & 0 & 1 & 0 & 0 \\
    0 & 0 & 0 & 1 & 0 \\
    1/4 & 3/4 & 0 & 0 & 0 
\end{pmatrix}$$

b) Utilice tecnología para encontrar $\lim_{n\to\infty} P^{n}$. Explique el comportamiento a largo plazo de la cadena.

Se utilizará la libraría `NumPy` en su moódulo `linalg` qur tiene integrada la función `matrix_power`, la cual eleva una matriz a una potencia dada. Tambien se utiliza la librería `SymPy` para una visualización más estética de las matrices.

El aborde que se le da al problema será el siguiente: Se tomará la matriz de trancisión, luego se elevará a potencias cada vez mayores, hasta un máximo de, por ejemplo, $2^7=128$. En cada iteración de la potencia se mostrará la matriz resultante para su comparación cn el resto. Es decir, se calculará $P^2, P^4, P^8, \dotsc, P^{128}$ y se mostrará cada una, para poder compararlas entre sí.

```python
import numpy as np
from sympy import Matrix, init_printing

# Para que SymPy se vea bien en el notebook
init_printing(use_unicode=True)

P = np.array(
    [
        [ 0 , 1/4 , 0 , 0 , 3/4 ],
        [ 3/4 , 0 , 0 , 0 , 1/4 ],
        [ 0 , 0 , 1 , 0 , 0 ],
        [ 0 , 0 , 0 , 1 , 0 ],
        [ 1/4 , 3/4 , 0 , 0 , 0 ]
    ]
, dtype = float)
    
print(f"--- Matriz Original (P^1) ---")
display(Matrix(np.round(P, 4)))

# hacemos el bucle para cada iteracion
for i in range(1, 8):
    potencia = 2**i
    
    # Elevamos al cuadrado el resultado anterior (P^8=(P^4)^2, P^4=(P^2)^2...)
    P = np.linalg.matrix_power(P, 2)
    
    print(f"\n--- Matriz Resultante (P^{potencia}) ---")

    # Redondeamos a 4 decimales para que no se vea caótico
    matriz_bonita = Matrix(np.round(P, 4))
    display(matriz_bonita)

```

Se puede ver que a partir de $P^{32}$ el resultado ya no cambia (a 4 cifras decimales), por lo que ahora se mostrarán las matrices cada 2 pasos de $P^{16}$ hasta $P^{32}$, es decir, se mostrarán $P^{16},P^{18},P^{20},\dotsc,P^{32}$, esto pues $P^{16}$ es la última matriz que vimos antes de que los 4 decimales ya no cambien, y $P^{32}$ es la primera que vimos a partir de la cual los 4 decimales ya no cambian

```python
import numpy as np
from sympy import Matrix, init_printing

# Para que SymPy se vea bien en el notebook
init_printing(use_unicode=True)

P = np.array(
    [
        [ 0 , 1/4 , 0 , 0 , 3/4 ],
        [ 3/4 , 0 , 0 , 0 , 1/4 ],
        [ 0 , 0 , 1 , 0 , 0 ],
        [ 0 , 0 , 0 , 1 , 0 ],
        [ 1/4 , 3/4 , 0 , 0 , 0 ]
    ]
, dtype = float)
    
print(f"--- Matriz Original (P^1) ---")
display(Matrix(np.round(P, 4)))

# hacemos el bucle para cada iteracion
for potencia in range(16, 34, 2):

    # Elevamos al cuadrado el resultado anterior (P^8=(P^4)^2, P^4=(P^2)^2...)
    P_i = np.linalg.matrix_power(P, potencia)
    
    print(f"\n--- Matriz Resultante (P^{potencia}) ---")

    # Redondeamos a 4 decimales para que no se vea caótico
    matriz_bonita = Matrix(np.round(P_i, 4))
    display(matriz_bonita)

```

Ahora vemos que desde $P^{24}$ los dígitos ya son muy parecidos, y a partir de $P^{26}$ ya no cambian. Esto indica, que a partir de 26 pasos de la cadena de Markov, se vuelve estacionaria entre estos grupos:

- Grupo 1: Estado 3 (absorbente)
- Grupo 2: Estado 4 (absorbente)
- Grupo 3: Estados 1, 2, 5 (recurrentes)
