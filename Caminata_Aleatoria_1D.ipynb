En una notebook de Jupyter, escribir un código que implemente una caminata aleatoria 1D, con probabilidad p de moverse a la derecha y probabilidad q de moverse a la izquierda, tal que p + q = 1. 

Una caminata aleatoria se define como el proceso: 
$$X_t = \sum_{i=1}^{t} B_i$$
donde $B_i$ es una variable aleatoria con la siguiente distribución de probabilidad:
$$P(B=k) = \begin{cases}
p & \text{si } & k = 1 \\
1-p & \text{si } & k = -1\\
0 & \text{en otro caso} & 
\end{cases}$$


Una caminata aleatoria recibe su nombre debido a que se puede pensar en el proceso como una serie de pasos que se dan sobre la recta numérica, ya sea a la izquierda o a la derecha con sus respectivas probabilidades, iniciando desde el origen.
Visto así, el proceso $X_t$ representa la posición de la persona que camina en el tiempo $t$, y la sucesión $0,X_1, X_2, X_3, \dotsc, X_t$ representa la trayectoria de la persona.

```python
import numpy as np
import matplotlib.pyplot as plt

p = 1.5 # Valor inicial no válido
# Ingresar valor de p
while not (p <= 1 and p >= 0):
    p = float(input("Escriba el valor de p\t"))
    if p < 0 or p > 1:
        print("El valor debe estar en [0,1]")

# Elegí 100 pasos de forma arbitraria
def caminata(p):
    n = 100
    trayectoria = np.empty(100, dtype=int)
    x_i = 0
    for i in range(n):
        b_i = np.random.uniform(0,1)
        if b_i <= p:
            x_i += 1
        else:
            x_i -= 1
        trayectoria[i] = x_i
    return trayectoria # Devuelve la lista de 100 pasos (trayectoria)


x_t = caminata(p)
linea_0 = np.zeros(100)

plt.style.use('classic')
plt.figure(figsize = (12, 6))
plt.plot(x_t, color = '#800040', linewidth = 3) # Color guinda
plt.plot(linea_0, color = 'black', linewidth = 0.5)
plt.title('Caminata Aleatoria (p = %4.2f)'%p, fontsize = 20, fontweight = 'bold')
plt.xlabel('Tiempo (# Pasos)', fontsize = 14)
plt.ylabel('Posición', fontsize = 14)
plt.grid(True)
plt.show()
```

## Análisis y Conclusiones
Al probar distintos valores de $p$, se puede notar que:
1. Si $p = \frac{1}{2}$, no hay un patrón claro, es el valor que parece más caótico
2. Si $p < \frac{1}{2}$, la trayectoria tiende a ir hacia abajo, en el extremo donde $p = 0$ la trayectoria se vuelve la recta $y = -x$
3. Si $p > \frac{1}{2}$, la trayectoria tiende a ir hacia abajo, en el extremo donde $p = 1$ la trayectoria se vuelve la recta $y = x$

Es importante destacar que se trata de un proceso aleatorio, por lo que al repetir el código varias veces, incluso con el mismo valor de $p$, se obtienen trayectorias distintas. Este experimento se realiza a continuación, se grafican sobrepuestas 5 trayectorias posibles con el mismo valor de $p$:

```python
import numpy as np
import matplotlib.pyplot as plt

p = 1.5
# Ingresar valor de p
while not (p <= 1 and p >= 0):
    p = float(input("Escriba el valor de p\t"))
    if p < 0 or p > 1:
        print("El valor debe estar en [0,1]")

# Elegí 100 pasos de forma arbitraria
def caminata(p):
    n = 100
    trayectoria = np.empty(100, dtype=int)
    x_i = 0
    for i in range(n):
        b_i = np.random.uniform(0,1)
        if b_i <= p:
            x_i += 1
        else:
            x_i -= 1
        trayectoria[i] = x_i
    return trayectoria # Devuelve la lista de 100 pasos (trayectoria)

plt.style.use('classic')
plt.figure(figsize = (12, 6))
plt.title('Caminatas Aleatorias (p = %4.2f)'%p, fontsize = 20, fontweight = 'bold')
plt.xlabel('Tiempo (# Pasos)', fontsize = 14)
plt.ylabel('Posición', fontsize = 14)
plt.grid(True)
x_t = np.empty((5,100))
for i in range(5):
    x_t[i,:] = caminata(p)
    plt.plot(x_t[i,:], linewidth = 2) # Color guinda
plt.show()

```
